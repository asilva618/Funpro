```python
notas = []
for i in range(5):
  nota = float(input(f"Ingrese la nota {i+1}: "))
  notas.append(nota)
# Encontrar la nota más baja
nota_mas_baja = min(notas)
# Eliminar la nota más baja de la lista
notas.remove(nota_mas_baja)
# Calcular el promedio de las notas restantes
promedio = sum(notas) / len(notas)
print(f"Las notas ingresadas son: {notas}")
print(f"La nota más baja es: {nota_mas_baja}")
print(f"El promedio de las notas restantes es: {promedio:.2f}")
```
