```python
# Ejemplo de lista de notas (escala de 0 a 5)
notas = [3.5, 4.0, 5.0, 2.5, 3.0, 4.5, 3.8, 3.9]

# Función para calcular el promedio
def calcular_promedio(lista_notas):
    # Si la lista está vacía, el promedio es 0 para evitar errores
    if not lista_notas:
        return 0
    suma_total = 0
    for nota in lista_notas:
        suma_total = suma_total + nota
    return suma_total / len(lista_notas)

# Calcular y mostrar la nota máxima
nota_maxima = max(notas)
print(f"Nota Máxima: {nota_maxima}")

# Calcular y mostrar la nota mínima
nota_minima = min(notas)
print(f"Nota Mínima: {nota_minima}")

# Calcular y mostrar el promedio
promedio = calcular_promedio(notas)
print(f"Promedio de Notas: {promedio:.2f}")

# Contar cuántas notas están por encima de 3 con un bucle
contador_notas_altas = 0
for nota in notas:
    if nota > 3:
        contador_notas_altas = contador_notas_altas + 1
print(f"Notas por encima de 3: {contador_notas_altas}")
```

```python
def analizar_notas(lista_notas):
    n_notas = len(lista_notas)
    nota_max = max(lista_notas)
    nota_min = min(lista_notas)
    i = 0
    sobre_tres = 0

    while i < n_notas:
        if lista_notas[i] >= 3:
            sobre_tres += 1
        i += 1

    print("Nota máxima:", nota_max)
    print("Nota mínima:", nota_min)
    print("Cantidad de notas sobre 3:", sobre_tres)

    return nota_max, nota_min, sobre_tres

notas = [4,4.5,3.5] #dentro del parentesis escriba la Nota 1, Nota 2, etc
analizar_notas(notas)
```
