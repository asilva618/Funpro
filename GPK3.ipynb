```python
import random
import time

# --- DATOS DEL JUGADOR ---
jugador = {
    "nombre": input("👤 Ingresa el nombre de tu héroe: "),
    "nivel": 1,
    "experiencia": 0,
    "inventario": []
}

# --- ENEMIGOS DISPONIBLES ---
enemigos = [
    {"nombre": "Goblin", "nivel": 1, "vida": 30, "exp": 20},
    {"nombre": "Orco", "nivel": 2, "vida": 50, "exp": 40},
    {"nombre": "Dragón", "nivel": 3, "vida": 100, "exp": 80}
]

# --- OBJETOS POSIBLES ---
objetos = ["Poción de vida", "Espada mágica", "Escudo antiguo"]

# --- FUNCIÓN PARA ELEGIR UN ENEMIGO ALEATORIO ---
def enemigo_aleatorio():
    return random.choice(enemigos)

# --- FUNCIÓN PARA EL DROP DE OBJETOS ---
def drop_objeto():
    if random.random() < 0.75:  # 75% de probabilidad de que dropee algo
        objeto = random.choice(objetos)
        jugador["inventario"].append(objeto)
        print(f"🎁 ¡El enemigo ha soltado un {objeto}!")
    else:
        print("💨 El enemigo no dejó nada...")

# --- FUNCIÓN DE BATALLA ---
def batalla():
    enemigo = enemigo_aleatorio()
    print(f"\n⚔️ ¡Un {enemigo['nombre']} (Nivel {enemigo['nivel']}) aparece con {enemigo['vida']} de vida!")

    # Simular batalla simple
    while enemigo["vida"] > 0:
        daño = random.randint(10, 30)
        enemigo["vida"] -= daño
        print(f"💥 Atacas y haces {daño} de daño. Vida restante del enemigo: {max(0, enemigo['vida'])}")
        time.sleep(1)

    print(f"🏆 ¡Has derrotado al {enemigo['nombre']}!")

    # Experiencia ganada
    jugador["experiencia"] += enemigo["exp"]
    print(f"✨ Has ganado {enemigo['exp']} puntos de experiencia. Total: {jugador['experiencia']} XP")

    drop_objeto()

# --- INICIO DEL JUEGO ---
print("\n🌟 ¡Bienvenido a la Aventura de Rol, valiente héroe! 🌟")

while True:
    accion = input("\n¿Quieres luchar contra un enemigo? (s/n): ").lower()
    if accion == "s":
        batalla()
    else:
        print(f"\n👋 Fin de la aventura, {jugador['nombre']}!")
        print(f"Nivel alcanzado: {jugador['nivel']} | Experiencia total: {jugador['experiencia']}")
        print(f"Inventario: {jugador['inventario']}")
        break
```
